```python
import numpy as np
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import accuracy_score, mean_absolute_error, r2_score


# Reproducible random results
rng = np.random.default_rng(42)


# ==========================================
# 1. FAULT COMBINATION READY KARO
# ==========================================

fault_level_names = le_fault_level.inverse_transform(
    df["Fault_Level"].astype(int)
)

fault_names = le_fault.inverse_transform(
    df["Most_Likely_Fault"].astype(int)
)

df["Fault_Combination"] = (
    pd.Series(fault_level_names, index=df.index)
    + " - "
    + pd.Series(fault_names, index=df.index)
)


# Combination -> encoded values mapping
fault_mapping = {}

for combination, level_code, fault_code in zip(
    df["Fault_Combination"],
    df["Fault_Level"],
    df["Most_Likely_Fault"]
):
    fault_mapping[combination] = {
        "Fault_Level": int(level_code),
        "Most_Likely_Fault": int(fault_code)
    }


all_combinations = list(fault_mapping.keys())


# Original recovery-time median preserve karo
combination_median_time = (
    df.groupby("Fault_Combination")[
        "Actual_Restoration_Min"
    ]
    .median()
    .to_dict()
)

global_median_time = df[
    "Actual_Restoration_Min"
].median()


# ==========================================
# 2. AREA-SPECIFIC FAULT PATTERN BANAO
# ==========================================

for area_code in df["Main_Area"].unique():

    area_mask = df["Main_Area"] == area_code
    area_indices = df.index[area_mask]

    # Is area ke existing top 3 faults
    existing_top = (
        df.loc[area_indices, "Fault_Combination"]
        .value_counts()
        .head(3)
        .index
        .tolist()
    )

    # Safety: at least 3 combinations
    remaining_combinations = [
        fault for fault in all_combinations
        if fault not in existing_top
    ]

    while len(existing_top) < 3:
        existing_top.append(
            remaining_combinations.pop(0)
        )

    other_faults = [
        fault for fault in all_combinations
        if fault not in existing_top
    ]

    # Top 3 receive 90%; remaining receive 10%
    choices = existing_top + other_faults

    probabilities = [
        0.50,   # Dominant fault
        0.25,   # Second fault
        0.15    # Third fault
    ]

    if other_faults:
        remaining_probability = 0.10 / len(other_faults)

        probabilities += [
            remaining_probability
        ] * len(other_faults)

    # Numerical safety
    probabilities = np.array(probabilities)
    probabilities = probabilities / probabilities.sum()

    assigned_faults = rng.choice(
        choices,
        size=len(area_indices),
        p=probabilities
    )

    df.loc[
        area_indices,
        "Fault_Combination"
    ] = assigned_faults


# ==========================================
# 3. FAULT CODES AND RECOVERY TIME UPDATE
# ==========================================

for index in df.index:

    combination = df.at[
        index,
        "Fault_Combination"
    ]

    mapping = fault_mapping[combination]

    df.at[
        index,
        "Fault_Level"
    ] = mapping["Fault_Level"]

    df.at[
        index,
        "Most_Likely_Fault"
    ] = mapping["Most_Likely_Fault"]

    base_time = combination_median_time.get(
        combination,
        global_median_time
    )

    # Recovery-time variation
    noise = rng.normal(
        loc=0,
        scale=max(10, base_time * 0.12)
    )

    recovery_minutes = round(
        base_time + noise
    )

    df.at[
        index,
        "Actual_Restoration_Min"
    ] = max(15, recovery_minutes)


# ==========================================
# 4. CHECK ARYA SAMAJ ROAD DISTRIBUTION
# ==========================================

arya_code = le_area.transform(
    ["Arya Samaj Road"]
)[0]

arya_distribution = (
    df[df["Main_Area"] == arya_code][
        "Fault_Combination"
    ]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Arya Samaj Road distribution:")
print(arya_distribution.head(5))


# ==========================================
# 5. AREA -> FAULT MODEL
# ==========================================

X_fault = df[["Main_Area"]]
y_fault = df["Fault_Combination"]

X_train_f, X_test_f, y_train_f, y_test_f = (
    train_test_split(
        X_fault,
        y_fault,
        test_size=0.2,
        random_state=42,
        stratify=y_fault
    )
)

area_fault_model = DecisionTreeClassifier(
    random_state=42
)

area_fault_model.fit(
    X_train_f,
    y_train_f
)

fault_predictions = area_fault_model.predict(
    X_test_f
)

print(
    "\nFault accuracy:",
    accuracy_score(
        y_test_f,
        fault_predictions
    )
)


# Final model ko complete dataset par fit karo
area_fault_model.fit(
    X_fault,
    y_fault
)


# ==========================================
# 6. RECOVERY-TIME MODEL
# ==========================================

X_time = df[[
    "Main_Area",
    "Fault_Level",
    "Most_Likely_Fault"
]]

y_time = df["Actual_Restoration_Min"]

X_train_t, X_test_t, y_train_t, y_test_t = (
    train_test_split(
        X_time,
        y_time,
        test_size=0.2,
        random_state=42
    )
)

recovery_model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

recovery_model.fit(
    X_train_t,
    y_train_t
)

time_predictions = recovery_model.predict(
    X_test_t
)

print(
    "Recovery MAE:",
    mean_absolute_error(
        y_test_t,
        time_predictions
    )
)

print(
    "Recovery R2:",
    r2_score(
        y_test_t,
        time_predictions
    )
)


# Final model complete dataset par fit
recovery_model.fit(
    X_time,
    y_time
)


# ==========================================
# 7. SAVE UPDATED FILES
# ==========================================

joblib.dump(
    area_fault_model,
    "area_fault_model.pkl"
)

joblib.dump(
    recovery_model,
    "recovery_model.pkl"
)

joblib.dump(
    le_area,
    "area_encoder.pkl"
)

joblib.dump(
    fault_mapping,
    "fault_mapping.pkl"
)

df.to_csv(
    "Muzaffarnagar_Realistic_Fault_Dataset.csv",
    index=False
)

print("\nUpdated dataset and models saved")
```

```python
import numpy as np
import pandas as pd
import joblib

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor


rng = np.random.default_rng(100)

all_areas = sorted(
    df["Main_Area"].unique()
)

all_faults = sorted(
    fault_mapping.keys()
)

number_of_faults = len(all_faults)


# =====================================
# DIFFERENT PROFILE FOR EVERY AREA
# =====================================

for area_position, area_code in enumerate(all_areas):

    area_indices = df.index[
        df["Main_Area"] == area_code
    ]

    # Different dominant faults for each area
    first_fault = all_faults[
        area_position % number_of_faults
    ]

    second_fault = all_faults[
        (area_position + 5) % number_of_faults
    ]

    third_fault = all_faults[
        (area_position + 10) % number_of_faults
    ]

    dominant_faults = [
        first_fault,
        second_fault,
        third_fault
    ]

    other_faults = [
        fault for fault in all_faults
        if fault not in dominant_faults
    ]

    choices = dominant_faults + other_faults

    probabilities = [
        0.50,  # Area ka main fault
        0.25,  # Second fault
        0.15   # Third fault
    ]

    if other_faults:

        other_probability = (
            0.10 / len(other_faults)
        )

        probabilities += [
            other_probability
        ] * len(other_faults)

    probabilities = np.array(probabilities)
    probabilities /= probabilities.sum()

    assigned_faults = rng.choice(
        choices,
        size=len(area_indices),
        p=probabilities
    )

    df.loc[
        area_indices,
        "Fault_Combination"
    ] = assigned_faults


# =====================================
# UPDATE FAULT CODES AND TIME
# =====================================

for index in df.index:

    combination = df.at[
        index,
        "Fault_Combination"
    ]

    mapping = fault_mapping[combination]

    level_code = mapping["Fault_Level"]
    fault_code = mapping["Most_Likely_Fault"]

    df.at[
        index,
        "Fault_Level"
    ] = level_code

    df.at[
        index,
        "Most_Likely_Fault"
    ] = fault_code

    base_time = combination_median_time.get(
        combination,
        global_median_time
    )

    noise = rng.normal(
        0,
        max(10, base_time * 0.12)
    )

    df.at[
        index,
        "Actual_Restoration_Min"
    ] = max(
        15,
        round(base_time + noise)
    )


# =====================================
# SHOW EACH AREA'S MAIN FAULT
# =====================================

print("Area-wise dominant faults:\n")

for area_code in all_areas:

    area_name = le_area.inverse_transform(
        [int(area_code)]
    )[0]

    area_distribution = (
        df[df["Main_Area"] == area_code][
            "Fault_Combination"
        ]
        .value_counts(normalize=True)
        .mul(100)
    )

    print(
        f"{area_name}: "
        f"{area_distribution.index[0]} = "
        f"{area_distribution.iloc[0]:.2f}%"
    )


# =====================================
# RETRAIN FAULT MODEL
# =====================================

X_fault = df[["Main_Area"]]
y_fault = df["Fault_Combination"]

area_fault_model = DecisionTreeClassifier(
    random_state=42
)

area_fault_model.fit(
    X_fault,
    y_fault
)


# =====================================
# RETRAIN RECOVERY MODEL
# =====================================

X_time = df[[
    "Main_Area",
    "Fault_Level",
    "Most_Likely_Fault"
]]

y_time = df["Actual_Restoration_Min"]

recovery_model = RandomForestRegressor(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

recovery_model.fit(
    X_time,
    y_time
)


# =====================================
# SAVE UPDATED FILES
# =====================================

joblib.dump(
    area_fault_model,
    "area_fault_model.pkl"
)

joblib.dump(
    recovery_model,
    "recovery_model.pkl"
)

joblib.dump(
    le_area,
    "area_encoder.pkl"
)

joblib.dump(
    fault_mapping,
    "fault_mapping.pkl"
)

df.to_csv(
    "Muzaffarnagar_Realistic_Fault_Dataset.csv",
    index=False
)

print("\nDifferent area profiles saved successfully")
```
